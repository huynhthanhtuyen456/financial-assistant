```python
import pandas as pd
import numpy as np
import requests
```

```python
yearly = True
response = requests.get(f'http://localhost/api/v1/scfa/cashflow?yearly={yearly}')
cashflow = response.json()
cashflow
```

```python
# Convert cashflow dictionary to pandas DataFrame
rows = []
for year, companies in cashflow.items():
    for symbol, metrics in companies.items():
        row = {'year': year, 'symbol': symbol}
        row.update(metrics)
        rows.append(row)

df_cashflow = pd.DataFrame(rows)
df_cashflow = df_cashflow.sort_values(['symbol', 'year']).reset_index(drop=True)
df_cashflow = df_cashflow.fillna(value=0)
if not yearly:
    df_cashflow['quarter'] = df_cashflow['year'].str.split('-Q').str[1].astype(int)
df_cashflow['year'] = df_cashflow['year'].str.split('-').str[0].astype(int)
symbols = df_cashflow['symbol'].unique()
```

```python
numberic_cols = df_cashflow.select_dtypes(include=[np.number]).columns.tolist()
numberic_cols
```

```python
df_cashflow.head()
```

```python
df_cashflow.sort_values(by=['symbol', 'year'], inplace=True)
df_cashflow.reset_index(drop=True, inplace=True)
```

# Discussion of df_cashflow

## Overview
- Size: 22,147 rows × 7 columns.
- Time span: annual data covering ~2000–2024 (dataset built from yearly API responses).
- Entities: ~1,582 unique symbols.
- Granularity: yearly (variable `year` is an integer); quarterly parsing is available in code but not used here (`yearly = True`).
- Missing values: NaNs were replaced with 0 (dataset was filled with zeros).

## Columns and types
- year (int64) — calendar year.
- symbol (object) — company identifier.
- investCost, fromInvest, fromFinancial, fromSale, freeCashFlow (int64) — cash flow numeric metrics (signed values; inflows/outflows).

## Key characteristics & quality notes
- All numeric columns are integer typed and have no nulls after fill; check whether zeros represent true zeros or originally-missing values (could bias aggregates).
- Wide value range and clear outliers exist (companies with very large positive or negative cash flows); consider robust statistics or winsorization when aggregating.
- Most features are correlated to varying degrees (see correlation results below).

## Correlation insights (computed on the numeric features)
- investCost ↔ fromInvest: strong positive correlation ≈ 0.69 — companies that report higher investment outflows tend to also show related movement in `fromInvest`.
- investCost ↔ fromFinancial: notable negative correlation ≈ -0.55 — investment outflows tend to associate with opposite sign movement in financing flows for many companies.
- fromFinancial shows negative correlations with investCost and fromInvest and weak correlation with fromSale.
- freeCashFlow has low-to-moderate correlations with other features (example: ≈ 0.28 with investCost), indicating it partly captures independent operational results.

## Suggested next steps / analyses
- Visualize time trends:
    - Per-symbol time series for key metrics (line charts) for selected large/representative companies.
    - Aggregate trends (median/mean or robust quantiles) across all companies per year.
- Outlier handling:
    - Inspect top/bottom contributors by metric; decide on winsorizing, log transforms (for positive flows), or separate analysis for extreme cases.
- Panel / time-series analysis:
    - Seasonal or structural-change detection (e.g., rolling means, breakpoint detection).
    - Panel regressions to relate freeCashFlow to other flows controlling for company fixed effects.
- Dimensionality & segmentation:
    - PCA or clustering on the five numeric features to find groups of similar cash-flow profiles.
- Diagnostics:
    - Re-check how NaNs were filled — verify the API semantics (missing vs. zero).
    - Validate units and signs (ensure consistency across years and companies).

## Quick interpretive notes
- The dataset is well-suited for cross-sectional and panel analyses of corporate cash-flow behavior across many years and companies.
- Correlations suggest meaningful relationships between investment activity and financing behavior; combine with balance-sheet or income-statement data (if available) for richer inference.
- Watch for reporting artifacts and outliers when producing summary statistics or models.

```python
import plotly.graph_objects as go
import plotly.express as px
from plotly.subplots import make_subplots
```

```python
# Get balance sheet data for correlation analysis
# Select multiple features for correlation
correlation_features = [col for col in numberic_cols if col not in ['year', 'quarter']]

# Filter data to include only rows with complete feature data
df_corr = df_cashflow[correlation_features].copy()

# Calculate correlation matrix
correlation_matrix = df_corr.corr()

# Create interactive heatmap using Plotly
fig = go.Figure(data=go.Heatmap(
    z=correlation_matrix.values,
    x=correlation_matrix.columns,
    y=correlation_matrix.columns,
    colorscale='RdBu',
    zmid=0,
    text=correlation_matrix.values,
    texttemplate='%{text:.2f}',
    textfont={"size": 12},
    colorbar=dict(title="Correlation")
))

fig.update_layout(
    title='Correlation Matrix - Balance Sheet Features',
    xaxis_title='Features',
    yaxis_title='Features',
    width=700,
    height=600,
    xaxis={'side': 'bottom'},
    yaxis={'autorange': 'reversed'}
)

fig.show()
```

```python
# Calculate average currentRatio by symbol (across all years)
avg_cr_by_symbol = df_cashflow.groupby('symbol')['freeCashFlow'].mean().reset_index()
avg_cr_by_symbol.columns = ['symbol', 'avg_freeCashFlow']
avg_cr_by_symbol = avg_cr_by_symbol.sort_values('avg_freeCashFlow', ascending=False)

# Get top 50 highest and lowest
top_50_highest = avg_cr_by_symbol.nlargest(50, 'avg_freeCashFlow')
top_50_lowest = avg_cr_by_symbol.nsmallest(50, 'avg_freeCashFlow')

# Create subplots: 1 row, 2 columns
fig = make_subplots(
    rows=1, cols=2,
    subplot_titles=("Top 50 Highest Average Free Cash Flow", "Top 50 Lowest Average Free Cash Flow"),
    specs=[[{"type": "bar"}, {"type": "bar"}]]
)

# Add top 50 highest chart
fig.add_trace(
    go.Bar(
        x=top_50_highest['avg_freeCashFlow'],
        y=top_50_highest['symbol'],
        orientation='h',
        marker=dict(color='green', opacity=0.7),
        name='Highest',
        hovertemplate='<b>%{y}</b><br>Avg Free Cash Flow: %{x:.2f}<extra></extra>'
    ),
    row=1, col=1
)

# Add top 50 lowest chart
fig.add_trace(
    go.Bar(
        x=top_50_lowest['avg_freeCashFlow'],
        y=top_50_lowest['symbol'],
        orientation='h',
        marker=dict(color='red', opacity=0.7),
        name='Lowest',
        hovertemplate='<b>%{y}</b><br>Avg Free Cash Flow: %{x:.2f}<extra></extra>'
    ),
    row=1, col=2
)

# Update layout
fig.update_layout(
    title_text="Distribution of Average Free Cash Flow by Symbol (2000-2024)",
    height=800,
    width=1400,
    showlegend=False,
    hovermode='closest'
)

fig.update_xaxes(title_text="Average Free Cash Flow", row=1, col=1)
fig.update_xaxes(title_text="Average Free Cash Flow", row=1, col=2)
fig.update_yaxes(title_text="Symbol", row=1, col=1)
fig.update_yaxes(title_text="Symbol", row=1, col=2)

fig.show()

# Print summary statistics
print("Top 50 Highest Average Free Cash Flow:")
print(top_50_highest.to_string(index=False))
print(f"\nMean: {top_50_highest['avg_freeCashFlow'].mean():.2f}")
print(f"Median: {top_50_highest['avg_freeCashFlow'].median():.2f}")
print(f"Std: {top_50_highest['avg_freeCashFlow'].std():.2f}")

print("\n" + "="*80 + "\n")

print("Top 50 Lowest Average Free Cash Flow:")
print(top_50_lowest.to_string(index=False))
print(f"\nMean: {top_50_lowest['avg_freeCashFlow'].mean():.2f}")
print(f"Median: {top_50_lowest['avg_freeCashFlow'].median():.2f}")
print(f"Std: {top_50_lowest['avg_freeCashFlow'].std():.2f}")
```

```python
yearly = True
response = requests.get(f'http://localhost/api/v1/scfa/balancesheet?yearly={yearly}')
balance_sheet = response.json()
balance_sheet
```

```python
# Convert balance_sheet dictionary to pandas DataFrame
rows = []
for year, companies in balance_sheet.items():
    for symbol, metrics in companies.items():
        row = {'year': year, 'symbol': symbol}
        row.update(metrics)
        rows.append(row)

df_balance_sheet = pd.DataFrame(rows)
df_balance_sheet = df_balance_sheet.sort_values(['symbol', 'year']).reset_index(drop=True)
df_balance_sheet = df_balance_sheet.fillna(value=0)
if not yearly:
    df_balance_sheet['quarter'] = df_balance_sheet['year'].str.split('-Q').str[1].astype(int)
df_balance_sheet['year'] = df_balance_sheet['year'].str.split('-').str[0].astype(int)
symbols = df_balance_sheet['symbol'].unique()
```

```python
df_cashflow["freeCashFlowToEquity"] = df_cashflow["fromSale"] - df_cashflow["investCost"] + df_balance_sheet["equity"]
```

```python
df_cashflow[['symbol', 'year', 'freeCashFlowToEquity']]
```
