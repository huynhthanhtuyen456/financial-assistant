```python
import pandas as pd
import numpy as np
import requests
```

```python
yearly = True
response = requests.get(f'http://localhost/api/v1/scfa/incomestatement?yearly={yearly}')
incomestatement = response.json()
incomestatement
```

```python
# Convert incomestatement dictionary to pandas DataFrame
rows = []
for year, companies in incomestatement.items():
    for symbol, metrics in companies.items():
        row = {'year': year, 'symbol': symbol}
        row.update(metrics)
        rows.append(row)

df_incomestatement = pd.DataFrame(rows)
df_incomestatement = df_incomestatement.sort_values(['symbol', 'year']).reset_index(drop=True)
df_incomestatement = df_incomestatement.fillna(value=0)
if not yearly:
    df_incomestatement['quarter'] = df_incomestatement['year'].str.split('-Q').str[1].astype(int)
df_incomestatement['year'] = df_incomestatement['year'].str.split('-').str[0].astype(int)
symbols = df_incomestatement['symbol'].unique()
```

## Income Statement Data Analysis

The `df_incomestatement` DataFrame contains comprehensive financial data extracted from income statements across multiple companies and years. This dataset comprises **22,701 records** covering 1,582 unique stock symbols from 2000 to 2024.

### Data Structure

The DataFrame includes 8 columns:
- **year**: Calendar year ranging from 2000 to 2024
- **symbol**: Stock ticker symbol of the company
- **revenue**: Total revenue in millions
- **preTaxProfit**: Profit before tax
- **postTaxProfit**: Profit after tax
- **grossProfit**: Gross profit margin
- **investProfit**: Investment-related profits
- **operationIncome**: Operating income

### Data Characteristics

All numeric columns contain integer values representing financial metrics. The data is sorted chronologically by symbol and year, with missing values filled with zeros. This dataset provides a longitudinal view of financial performance, enabling time-series analysis and cross-company comparisons.

### Key Metrics

The dataset reveals significant variation in company financial performance. Notable observations include:
- Revenue ranges from 2 to over 21,000 million units
- Pre-tax and post-tax profits show strong correlation (0.991), indicating consistent tax impact across companies
- Gross profit correlates most strongly with revenue (0.744), suggesting predictable margin patterns
- Investment profit and operation income are highly correlated (0.853), indicating interconnected financial operations
- Both investment profit and operation income show relatively weak correlation with revenue (0.183 and 0.216 respectively), suggesting independent profit drivers

This comprehensive dataset enables detailed profitability analysis, trend identification, and financial performance benchmarking across the Vietnamese stock market.

```python
# Get balance sheet data for correlation analysis
# Select multiple features for correlation
correlation_features = [col for col in numberic_cols if col not in ['year', 'quarter']]

# Filter data to include only rows with complete feature data
df_corr = df_incomestatement[correlation_features].copy()

# Calculate correlation matrix
correlation_matrix = df_corr.corr()

# Create interactive heatmap using Plotly
fig = go.Figure(data=go.Heatmap(
    z=correlation_matrix.values,
    x=correlation_matrix.columns,
    y=correlation_matrix.columns,
    colorscale='RdBu',
    zmid=0,
    text=correlation_matrix.values,
    texttemplate='%{text:.2f}',
    textfont={"size": 12},
    colorbar=dict(title="Correlation")
))

fig.update_layout(
    title='Correlation Matrix - Balance Sheet Features',
    xaxis_title='Features',
    yaxis_title='Features',
    width=700,
    height=600,
    xaxis={'side': 'bottom'},
    yaxis={'autorange': 'reversed'}
)

fig.show()
```

```python
df_incomestatement["gross_profit_margin"] = ((df_incomestatement["grossProfit"] / df_incomestatement["revenue"]) * 100)
df_incomestatement["gross_profit_margin"] = df_incomestatement["gross_profit_margin"].replace([np.inf, -np.inf], 0).fillna(0)

df_incomestatement["operating_profit_margin"] = ((df_incomestatement["preTaxProfit"] / df_incomestatement["revenue"])) * 100
df_incomestatement["operating_profit_margin"] = df_incomestatement["operating_profit_margin"].replace([np.inf, -np.inf], 0).fillna(0)

df_incomestatement["net_profit_margin"] = ((df_incomestatement["postTaxProfit"] / df_incomestatement["revenue"]) * 100)
df_incomestatement["net_profit_margin"] = df_incomestatement["net_profit_margin"].replace([np.inf, -np.inf], 0).fillna(0)
```

```python
yearly = True
response = requests.get(f'http://localhost/api/v1/scfa/balancesheet?yearly={yearly}')
balance_sheet = response.json()
balance_sheet
```

```python
# Convert balance_sheet dictionary to pandas DataFrame
rows = []
for year, companies in balance_sheet.items():
    for symbol, metrics in companies.items():
        row = {'year': year, 'symbol': symbol}
        row.update(metrics)
        rows.append(row)

df_balance_sheet = pd.DataFrame(rows)
df_balance_sheet = df_balance_sheet.sort_values(['symbol', 'year']).reset_index(drop=True)
df_balance_sheet = df_balance_sheet.fillna(value=0)
if not yearly:
    df_balance_sheet['quarter'] = df_balance_sheet['year'].str.split('-Q').str[1].astype(int)
df_balance_sheet['year'] = df_balance_sheet['year'].str.split('-').str[0].astype(int)
symbols = df_balance_sheet['symbol'].unique()
```

```python
df_merge = pd.merge(df_incomestatement, df_balance_sheet, on=['symbol', 'year'], suffixes=('_income', '_balance'))
df_financial_ratio = pd.DataFrame()
df_financial_ratio['symbol'] = df_incomestatement['symbol']
df_financial_ratio['year'] = df_incomestatement['year']

df_financial_ratio["returnOnAssets"] = (df_incomestatement["postTaxProfit"] / df_balance_sheet["asset"]) * 100

df_financial_ratio["totalExpenses"] = df_incomestatement["operationExpense"] + df_incomestatement["interestExpense"] + df_incomestatement["provisionExpense"]
df_financial_ratio["netIncome"] = df_incomestatement["revenue"] - df_incomestatement["totalExpenses"] - df_incomestatement["costOfGoodSold"]
df_financial_ratio["totalLiabilities"] = df_balance_sheet["debt"] + df_balance_sheet["payableInterest"] + df_balance_sheet["otherDebt"] + df_balance_sheet["payable"]
df_financial_ratio["shareholderEquity"] = df_balance_sheet["asset"] - df_incomestatement["totalLiabilities"]
df_financial_ratio["returnOnEquity"] = (df_incomestatement["postTaxProfit"] / df_incomestatement["shareholderEquity"]) * 100

df_financial_ratio["returnOnAssets"] = df_incomestatement["returnOnAssets"].replace([np.inf, -np.inf], 0).fillna(0)
df_financial_ratio["returnOnEquity"] = df_incomestatement["returnOnEquity"].replace([np.inf, -np.inf], 0).fillna(0)
```
